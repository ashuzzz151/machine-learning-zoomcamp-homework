```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
df.head()

# Q1: Pandas Version
print(f"Q1 Answer (Pandas Version): {pd.__version__}")

# Q2: Records Count
records_count = len(df)
print(f"Q2 Answer (Records Count): {records_count}")

# Q3: Fuel Types
fuel_types_count = df['fuel_type'].nunique()
print(f"Q3 Answer (Number of Fuel Types): {fuel_types_count}")

# Q4: Missing Values
cols_with_missing = (df.isnull().sum() > 0).sum()
print(f"Q4 Answer (Columns with Missing Values): {cols_with_missing}")

# Q5: Max Fuel Efficiency (using 'mpg')
max_asia_efficiency = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print(f"Q5 Answer (Max Fuel Efficiency for Asia): {max_asia_efficiency}")

# Q6: Median Value of Horsepower
initial_median = df['horsepower'].median()
hp_mode = df['horsepower'].mode()[0]

df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(hp_mode)
new_median = df_filled['horsepower'].median()

# Evaluate direction of change to match homework options
if new_median > initial_median:
    q6_answer = "Yes, it increased"
elif new_median < initial_median:
    q6_answer = "Yes, it decreased"
else:
    q6_answer = "No"

print(f"Initial Median: {initial_median}")
print(f"Most Frequent Value (Mode): {hp_mode}")
print(f"New Median: {new_median}")
print(f"Q6 Answer: {q6_answer}")
# Q7: Sum of Weights
asia_cars = df[df['origin'] == 'Asia']
asia_subset = asia_cars[['vehicle_weight', 'model_year']].iloc[:7]

X = asia_subset.values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = (XTX_inv @ X.T) @ y
w_sum = w.sum()

print(f"Weight vector w: {w}")
print(f"Q7 Answer (Sum of elements of w): {w_sum:.4f}")
```
